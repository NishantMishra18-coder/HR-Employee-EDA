```python

import pandas as pd

df = pd.read_csv('/content/cleaned_hr_employee_data.csv')

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

df.head()
```

```python

# Column names and data types
df.info()

# Missing values in each column
print("\nMissing values:")
print(df.isnull().sum())

# Duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# Statistical summary
df.describe()
```

```python

# Average employee age
print("Average Age:", round(df['Age'].mean(), 2))

# Average monthly income
print("Average Monthly Income:", round(df['MonthlyIncome'].mean(), 2))

# Department-wise employees
print("\nEmployees by Department:")
print(df['Department'].value_counts())

# Attrition count
print("\nAttrition Count:")
print(df['Attrition'].value_counts())

# Gender distribution
print("\nGender Distribution:")
print(df['Gender'].value_counts())

# Overtime distribution
print("\nOvertime Distribution:")
print(df['OverTime'].value_counts())
```

```python

import matplotlib.pyplot as plt

# 1. Employees by department
df['Department'].value_counts().plot(kind='bar')
plt.title('Employees by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.show()

# 2. Attrition distribution
df['Attrition'].value_counts().plot(kind='bar')
plt.title('Employee Attrition')
plt.xlabel('Attrition')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.show()

# 3. Monthly income distribution
df['MonthlyIncome'].plot(kind='hist', bins=20)
plt.title('Monthly Income Distribution')
plt.xlabel('Monthly Income')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()

# 4. Age distribution
df['Age'].plot(kind='hist', bins=15)
plt.title('Employee Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()

# 5. Overtime vs attrition
pd.crosstab(df['OverTime'], df['Attrition']).plot(kind='bar')
plt.title('Overtime vs Employee Attrition')
plt.xlabel('Overtime')
plt.ylabel('Number of Employees')
plt.tight_layout()
plt.show()
```

```python

# 1. Total employees
print("Total Employees:", len(df))

# 2. Department with the most employees
print("\nEmployees by Department:")
print(df['Department'].value_counts())

# 3. Attrition percentage
print("\nAttrition Percentage:")
print((df['Attrition'].value_counts(normalize=True) * 100).round(2))

# 4. Average age and monthly income
print("\nAverage Age:", round(df['Age'].mean(), 2))
print("Average Monthly Income:", round(df['MonthlyIncome'].mean(), 2))

# 5. Attrition percentage by overtime
print("\nAttrition Rate by Overtime:")
print((pd.crosstab(df['OverTime'], df['Attrition'], normalize='index') * 100).round(2))

# 6. Missing values and duplicate rows
print("\nTotal Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())
```
